```python
#Q2:

#1
import pandas as pd

df = pd.read_csv("A03-knn/data/land_mines.csv")
print(df.head())
print(df.describe())
print(df["mine_type"].value_counts())
```

```python
#2
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler

X = df[["voltage", "height", "soil"]]
y = df["mine_type"]

X_train, X_valid, y_train, y_valid = train_test_split(
    X, y,
    test_size=0.5,
    random_state=65
)

scaler = MinMaxScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_valid_scaled = scaler.transform(X_valid)
```

```python
#3
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

ks = np.arange(1, 31)
accuracies = []

for k in ks:
    model = KNeighborsClassifier(n_neighbors=k)
    model.fit(X_train_scaled, y_train)

    predictions = model.predict(X_valid_scaled)
    accuracies.append(accuracy_score(y_valid, predictions))

best_k = int(ks[np.argmax(accuracies)])

print("Best k:", best_k)
print("Best accuracy:", max(accuracies))
```

```python
#4
best_model = KNeighborsClassifier(n_neighbors=best_k)
best_model.fit(X_train_scaled, y_train)

predictions = best_model.predict(X_valid_scaled)

confusion = pd.crosstab(
    y_valid,
    predictions,
    rownames=["Actual"],
    colnames=["Predicted"]
)

print(confusion)
print("Accuracy:", accuracy_score(y_valid, predictions))
```

The confusion table shows the actual mine types in the rows and the predicted mine types in the columns. Values along the diagonal are correct predictions, while values outside the diagonal are mistakes. The model performs better for some mine types than others, which shows that overall accuracy does not tell the whole story.
